```python
#==============================================================================
# SISTEMA DE PRESTAMOS
#==============================================================================

#lista de alumnos activos en el sistema
alumnos = ["26040200", "26040201", "26040202", "26040203", "26040204"]

#Cada valor en la lista de computadoras tiene su lista correspondiente con el
#aula donde esta ubicada, el software instalado y el estado de esa computadora
computadoras = ["PC-LC1-01", "PC-LC1-02", "PC-LC1-03",
                "PC-LC2-01", "PC-LC2-02", "PC-LC2-03",
                "PC-LC3-01", "PC-LC3-02", "PC-LC3-03","PC-LC3-04"]

AULAS = ["LC1", "LC1", "LC1",
         "LC2", "LC2", "LC3",
         "LC3", "LC3", "LC3", "LC3"]
software = ["Pynthon, VS Code", "Pynthon, Java, VS Code", "AutoCAD, MATLAB",
            "Pynthon, VS Code", "Pynthon, Java, VS Code", "AutoCAD, MATLAB",
            "Pynthon, VS Code", "Pynthon, Java, VS Code", "AutoCAD, MATLAB", "Pynthon, VS Code"]
estados = [ "Disponible", "Prestada", "Reservada",
           "Disponible", "Prestada", "Reservada",
            "Disponible", "Prestada", "Reservada", "Mantenimiento"]

#MENU PRINCIPAL
#Usamos un ciclo While que se repetira indifinidamente hasta que el usuario
#teclee la opcion 7 (salir) o cualquier numero superior a 7
opcion = 0
while opcion <= 6:
    print("\n===============================")
    print("      SISTEMA DE PRESTAMOS")
    print("\n=============================")
    print("1. Consultar computadoras")
    print("2. Buscar por software")
    print("3. Solicitar prestamo")
    print("4. Devolver computadora")
    print("5. Crear reservacion")
    print("6. Agregar computadora computadora")
    print("7. Salir")
    print("\n============================")
    opcion = int(input("Seleccione una opcion: "))

    if opcion == 1:
        print("\n---COMPUTADORAS---")
        for i in range(len(computadoras)):
            print("Aula: ", AULAS[i],"ID: ", computadoras[i],"Software: ", software[i],"Estado: ", estados[i])
            print("\n-----------------------")
    elif opcion == 2:
        programa = input( "\nIngrese el software que desea buscar: ")
        print("\n---RESULTADOS---")
        encontrado = False
        for i in range(len(computadoras)):
            if programa.lower() in software[i].lower():
                encontrado = True
                print("ID:", computadoras[i])
                print("Aula:", AULAS[i])
                print("Software:", software[i])
                print("Estado:", estados[i])

        if encontrado == False:
            print("No se encontraron computadoras con ese software")
    elif opcion == 3:
        print("\n---SOLICITAR PRESTAMO---")
        num_control = input( "\nIngrese su numero de control: ")
        encontrado1 = False
        if num_control in alumnos:
            encontrado1 = True
            id_computadora = input("Ingrese el ID de la computadora: "  )
            encontrado2 = False
            for i in range(len(computadoras)):
                if computadoras[i] == id_computadora:
                    encontrado2 = True
                    if estados[i] == "Disponible":
                        estados[i] = "Prestada"
                        print("Prestamo registrado correctamente.")
                        print("Computadora:", computadoras[i])
                        print("Aula", AULAS[i])
                    else:
                        print("La computadora no esta disponible")
            if encontrado2 == False:
                print("La computadora no existe.")
        if encontrado1 == False:
          print("el alumno no existe.")
    elif opcion == 4:
        print("Opción 4: Devolver computadora (Funcionalidad no implementada aún).")
    elif opcion == 5:
        print("Opción 5: Crear reservación (Funcionalidad no implementada aún).")
    elif opcion == 6:
        print("Opción 6: Agregar computadora (Funcionalidad no implementada aún).")
    elif opcion == 7:
        print("Saliendo del sistema. ¡Hasta luego!")
        break
    else:
        print("Opción inválida. Por favor, seleccione una opción entre 1 y 7.")
```
